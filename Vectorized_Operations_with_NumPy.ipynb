```python
import numpy as np
```

```python
energy_data=np.array([[12, 14, 13, 15, 16, 14],
 [10, 11, 12, 13, 14, 12],
 [20, 22, 21, 23, 24, 22],
 [ 8,  9, 10, 11, 12, 10],
 [15, 16, 17, 18, 19, 17]
                      ])

print("Enery Data:")
print(energy_data)
print("Shape",energy_data.shape)
```

```python
adjustment_data=energy_data*1.10

print("Adjustment Data:")
print(adjustment_data)
```

```python
calibration_factors = [1.0, 0.98, 1.02, 1.01, 0.99, 1.03]

calibration_data=adjustment_data*calibration_factors

print("Calibration Data:")
print(calibration_data)
```

```python
Average_energy_per_machine=np.mean(calibration_data,axis=1)

print("Average Energy Per Machine:")
print(Average_energy_per_machine)
```

```python
Maximum_energy=np.max(calibration_data)

print("Maximum Energy:")
print(Maximum_energy)
```
